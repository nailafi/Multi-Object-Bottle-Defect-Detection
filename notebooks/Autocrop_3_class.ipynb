```python
import os
import cv2
import shutil
import numpy as np
from ultralytics import YOLO
```

```python
model = YOLO('/content/drive/MyDrive/best_last.pt')

#dataset asli
input_folder = '//content/drive/MyDrive/CAE_dataset_autocrop/CAE_tutup_normal_crop'

#output folder
label_folder = '/content/crop_labe'
tutup_folder   = '/content/crop_tutup'
botol_folder  = '/content/crop_boto'

os.makedirs(label_folder, exist_ok=True)
os.makedirs(tutup_folder, exist_ok=True)
os.makedirs(botol_folder, exist_ok=True)

#image size
label_size = 128
tutup_size   = 64
botol_size  = 128

#penambahan padding
padding = 5

#counter
count_label = 0
count_tutup   = 0
count_botol  = 0

#loop gambar
for filename in os.listdir(input_folder):
    if filename.endswith(('.jpg', '.png', '.jpeg')):
        img_path = os.path.join(
            input_folder,
            filename
        )
        img = cv2.imread(img_path)

        #YOLO detection
        results = model(img, conf=0.5)

        #loop detection
        for r in results:
            boxes = r.boxes
            for box in boxes:
                #class id
                class_id = int(box.cls[0])

                #bounding box
                x1, y1, x2, y2 = map(
                    int,
                    box.xyxy[0]
                )

                #padding
                x1 = max(0, x1 - padding)
                y1 = max(0, y1 - padding)

                x2 = min(img.shape[1], x2 + padding)
                y2 = min(img.shape[0], y2 + padding)

                #crop
                crop_img = img[y1:y2, x1:x2]

                if crop_img.size > 0:

                    #label = 1
                    if class_id == 1:

                        resized = cv2.resize(
                            crop_img,
                            (label_size, label_size)
                        )

                        save_path = os.path.join(
                            label_folder,
                            f'label_{count_label}.jpg'
                        )

                        cv2.imwrite(save_path, resized)

                        count_label += 1

                    #tutup = 2
                    elif class_id == 2:

                        resized = cv2.resize(
                            crop_img,
                            (tutup_size, tutup_size)
                        )

                        save_path = os.path.join(
                            tutup_folder,
                            f'tutup_{count_tutup}.jpg'
                        )

                        cv2.imwrite(save_path, resized)

                        count_tutup += 1

                    #botol
                    elif class_id == 0:

                        resized = cv2.resize(
                            crop_img,
                            (botol_size, botol_size)
                        )

                        save_path = os.path.join(
                            botol_folder,
                            f'body_{count_botol}.jpg'
                        )

                        cv2.imwrite(save_path, resized)

                        count_botol += 1

#results
print(f'Crop Label : {count_label}')
print(f'Crop Tutup : {count_tutup}')
print(f'Crop Body  : {count_botol}')

#simpan jadi ZIP
shutil.make_archive('dataset_label', 'zip', label_folder)
shutil.make_archive('dataset_tutup', 'zip', tutup_folder)
shutil.make_archive('dataset_botol', 'zip', botol_folder)

print('Semua dataset berhasil dibuat')
```
